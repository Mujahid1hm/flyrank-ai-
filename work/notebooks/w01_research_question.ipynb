## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
from pathlib import Path
import pandas as pd

search_roots = [Path.cwd(), *Path.cwd().parents]

def find_dataset():
    for root in search_roots:
        candidate = root / "data" / "raw" / "content_refresh_anonymized.csv"
        if candidate.exists():
            return candidate
    for root in search_roots:
        matches = list(root.rglob("content_refresh_anonymized.csv"))
        if matches:
            return matches[0]
    return None

candidate = find_dataset()

if candidate is not None:
    df = pd.read_csv(candidate)
    print("--- Starter Dataset Summary ---")
    print(f"1. Total Units of Analysis (Rows/Pages): {len(df):,}")
    print(f"2. Total Columns/Features: {len(df.columns)}")

    num_cols = df.select_dtypes(include=['number']).columns.tolist()
    if num_cols:
        first_num = num_cols[0]
        print(f"3. Average {first_num}: {df[first_num].mean():.2f}")
    else:
        print("3. Dataset preview:\n" + df.head(2).to_string(index=False))
else:
    print("Dataset path checked. Please ensure starter CSV is placed under data/ directory.")
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```
